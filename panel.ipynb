```python
#CARGAR LOS DATOS
import json

def cargar_ventas():
    with open("data.json", "r") as archivo:
        datos = json.load(archivo)

    return datos["ventas"]   
```

```python
#PRODUCTO CON MÁS INGRESOS
productos = {}

for venta in ventas:
    producto = venta["producto"]
    ingresos = venta["cantidad"] * venta["precio_unitario"]

    if producto in productos:
        productos[producto] = productos[producto] + ingresos
    else:
        productos[producto] = ingresos

producto_mas_ingresos = max(productos, key=productos.get)

print("Producto con más ingresos:", producto_mas_ingresos)
print("Ingresos:", productos[producto_mas_ingresos])
```

```python
#FILTRAR VENTANAS POR VENDEDOR CONCRETO
def filtrar_ventas(vendedor):
    ventas_vendedor = []

    for venta in ventas:
        if venta["vendedor"] == vendedor:
            ventas_vendedor.append(venta)

    return ventas_vendedor
```

```python
#Ahora haré una prueba filtrando un vendedor, por ejemplo: Luis
print ("FILTRO LUIS")
filtrar_ventas("Luis")
```

```python
#pruebo también: Ana
print ("FILTRO ANA")
filtrar_ventas("Ana")
```

```python
#CALCULAR TICKET MEDIO
total = 0

for venta in ventas:
    total = total + venta["cantidad"] * venta["precio_unitario"]

ticket_medio = total / len(ventas)

print("TICKET MEDIO:", ticket_medio)
```
